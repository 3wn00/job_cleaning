alot of columns are not standardized lets standerdize them

```python
def normalize(val):
    if pd.isna(val): return np.nan
    val = val.lower()
    if "part" in val: return "part_time"
    if "full" in val: return "full_time"
    if "contract" in val or "temp" in val: return "contract"
    return "other"

df["job_type_clean"] = df["job_type"].apply(normalize)
```

lets do them same for sector, i am going to classify the sectors in these categories:

```python
def classify_job(row):
    sector_val = str(row['sector']).lower() if pd.notna(row['sector']) else ''
    title_val = str(row['job_title']).lower() if pd.notna(row['job_title']) else ''
    desc_val = str(row['job_description']).lower()[:300] if pd.notna(row['job_description']) else ''
    
    # Combined text for sector inference (title most reliable, desc for backup)
    text = title_val + ' ' + desc_val

    seniority = np.nan
    if sector_val:
        if 'senior executive' in sector_val or 'president' in sector_val or 'cfo' in sector_val:
            seniority = 'Senior Executive'
        elif 'executive' in sector_val:
            seniority = 'Executive'
        elif 'manager' in sector_val:
            seniority = 'Manager'
        elif 'entry' in sector_val:
            seniority = 'Entry Level'
        elif 'experienced' in sector_val:
            seniority = 'Experienced (Non-Manager)'
        elif 'student' in sector_val:
            seniority = 'Student'
        elif 'intern' in sector_val:
            seniority = 'Intern'

    # SECTOR — try sector column first, fall back to title + description  #
    
    # If the sector value is a seniority label, skip it and go straight to text inference
    is_seniority_label = seniority is not np.nan
    
    sector = np.nan

    # --- Pass 1: from sector column (skip if it's a seniority label) ---
    if sector_val and not is_seniority_label:
        if any(x in sector_val for x in [
            'it/', 'software', 'hardware', 'network', 'database', 'developer',
            'systems admin', 'tech support', 'information tech', 'computer',
            'cyber', 'devops', 'cloud', 'qa/', 'quality assurance', 'telecom'
        ]):
            sector = 'IT & Software'

        elif any(x in sector_val for x in [
            'engineer', 'civil', 'structural', 'mechanical', 'electrical',
            'chemical', 'aerospace', 'industrial', 'packaging', 'petroleum'
        ]):
            sector = 'Engineering'

        elif any(x in sector_val for x in [
            'medical', 'health', 'nurs', 'pharma', 'lab/', 'laboratory',
            'dental', 'physician', 'therapy', 'therapist', 'clinical',
            'biotech', 'social work', 'mental health', 'optom', 'veterin',
            'radiol', 'pathol', 'surgeon', 'surgery'
        ]):
            sector = 'Healthcare'

        elif any(x in sector_val for x in [
            'sales', 'marketing', 'business dev', 'advertising',
            'public relations', 'brand', 'market research', 'e-commerce'
        ]):
            sector = 'Sales & Marketing'

        elif any(x in sector_val for x in [
            'financ', 'bank', 'insurance', 'invest', 'audit', 'tax',
            'payroll', 'actuar', 'mortgage', 'credit', 'wealth',
            'budget', 'controller', 'treasury', 'accountant', 'accounting'
        ]):
            sector = 'Finance & Accounting'

        elif any(x in sector_val for x in [
            'construct', 'maintenance', 'repair', 'install', 'electrician',
            'plumb', 'hvac', 'carpenter', 'weld', 'facilities', 'foreman'
        ]):
            sector = 'Construction & Trades'

        elif any(x in sector_val for x in [
            'logistic', 'transport', 'driver', 'warehouse', 'supply chain',
            'shipping', 'freight', 'delivery', 'fleet', 'aviation',
            'dispatch', 'procurement', 'purchasing'
        ]):
            sector = 'Logistics & Transport'

        elif any(x in sector_val for x in [
            'customer support', 'customer service', 'client care',
            'call center', 'reservations', 'help desk'
        ]):
            sector = 'Customer Service'

        elif any(x in sector_val for x in [
            'retail', 'food service', 'hospitality', 'restaurant',
            'hotel', 'travel', 'tourism', 'culinary', 'catering'
        ]):
            sector = 'Retail & Hospitality'

        elif any(x in sector_val for x in [
            'educat', 'teach', 'tutor', 'academic', 'curriculum',
            'school', 'university', 'faculty'
        ]):
            sector = 'Education'

        elif any(x in sector_val for x in [
            'human resource', ' hr/', 'recruiting', 'talent acqui',
            'admin', 'clerical', 'receptionist', 'office manage',
            'executive assist', 'data entry', 'secretar'
        ]):
            sector = 'Admin & HR'

        elif any(x in sector_val for x in [
            'legal', 'attorney', 'paralegal', 'compliance',
            'litigation', 'regulatory'
        ]):
            sector = 'Legal & Compliance'

        elif any(x in sector_val for x in [
            'project/program', 'program manag', 'project manag',
            'business/strategic', 'strategy', 'operations manag',
            'business analyst', 'scrum', 'agile'
        ]):
            sector = 'Project & Strategy'

        elif any(x in sector_val for x in [
            'design', 'creative', 'art direct', 'graphic', 'media',
            'journalism', 'writing', 'editor', 'content', 'publishing',
            'photography', 'broadcast'
        ]):
            sector = 'Creative & Media'

        elif any(x in sector_val for x in [
            'non-profit', 'nonprofit', 'government', 'military',
            'public sector', 'community'
        ]):
            sector = 'Non-profit & Government'

        elif any(x in sector_val for x in [
            'manufacturing', 'assembl', 'production',
            'quality control', 'lean', 'six sigma'
        ]):
            sector = 'Manufacturing'

        elif any(x in sector_val for x in [
            'security', 'protective', 'guard', 'surveillance'
        ]):
            sector = 'Security & Protective'

    #infer from job_title + job_description if still unknown
    if pd.isna(sector):
        if any(x in text for x in [
            'software', 'developer', 'devops', 'cloud', 'it support',
            'network admin', 'database', 'cybersecurity', 'systems admin',
            'data scientist', 'machine learning', 'backend', 'frontend',
            'full stack', 'helpdesk', 'web developer', 'matlab',
            'java ', 'python', '.net', 'sap', 'salesforce', 'erp'
        ]):
            sector = 'IT & Software'

        elif any(x in text for x in [
            'mechanical engineer', 'electrical engineer', 'civil engineer',
            'chemical engineer', 'aerospace engineer', 'structural engineer',
            'process engineer', 'quality engineer', 'packaging engineer',
            'environmental engineer', 'petroleum engineer'
        ]):
            sector = 'Engineering'

        elif any(x in text for x in [
            'nurse', 'physician', 'doctor', 'medical', 'clinical', 'dental',
            'pharmacy', 'therapist', 'healthcare', 'health care', 'surgeon',
            'caregiver', 'home health', 'patient', 'hospital', 'icu',
            'emt', 'paramedic', 'radiolog', 'patholog'
        ]):
            sector = 'Healthcare'

        elif any(x in text for x in [
            'accountant', 'accounting', 'financial analyst', 'finance manager',
            'auditor', 'tax ', 'payroll', 'bookkeeper', 'controller',
            'investment', 'mortgage', 'actuary', 'underwriter', 'credit analyst'
        ]):
            sector = 'Finance & Accounting'

        elif any(x in text for x in [
            'sales rep', 'sales manager', 'account executive', 'territory sales',
            'marketing manager', 'digital marketing', 'brand manager', 'seo',
            'social media', 'business development', 'sales support'
        ]):
            sector = 'Sales & Marketing'

        elif any(x in text for x in [
            'construction', 'electrician', 'plumber', 'hvac', 'carpenter',
            'welder', 'foreman', 'maintenance tech', 'facilities manager',
            'roofer', 'pipefitter', 'millwright', 'handyman'
        ]):
            sector = 'Construction & Trades'

        elif any(x in text for x in [
            'truck driver', 'cdl', 'warehouse', 'logistics', 'supply chain',
            'shipping', 'freight', 'dispatcher', 'forklift', 'delivery driver',
            'mailroom', 'procurement', 'purchasing', 'distribution'
        ]):
            sector = 'Logistics & Transport'

        elif any(x in text for x in [
            'customer service', 'customer support', 'call center',
            'client services', 'customer care', 'service rep'
        ]):
            sector = 'Customer Service'

        elif any(x in text for x in [
            'restaurant', 'chef', 'cook', 'bartender', 'server', 'hotel',
            'hospitality', 'food service', 'barista', 'cashier', 'retail',
            'store manager', 'shift supervisor'
        ]):
            sector = 'Retail & Hospitality'

        elif any(x in text for x in [
            'teacher', 'instructor', 'professor', 'tutor', 'educator',
            'curriculum', 'preschool', 'kindergarten', 'e-learning', 'faculty'
        ]):
            sector = 'Education'

        elif any(x in text for x in [
            'human resources', 'recruiter', 'talent acquisition',
            'office manager', 'administrative assistant', 'receptionist',
            'data entry', 'executive assistant', 'office admin'
        ]):
            sector = 'Admin & HR'

        elif any(x in text for x in [
            'attorney', 'lawyer', 'paralegal', 'legal', 'compliance officer',
            'contract admin', 'litigation', 'counsel'
        ]):
            sector = 'Legal & Compliance'

        elif any(x in text for x in [
            'project manager', 'program manager', 'scrum master', 'agile',
            'business analyst', 'product manager', 'product owner', 'pmo'
        ]):
            sector = 'Project & Strategy'

        elif any(x in text for x in [
            'graphic designer', 'ux designer', 'ui designer', 'copywriter',
            'journalist', 'content writer', 'art director', 'photographer',
            'videographer', 'animator', 'creative director'
        ]):
            sector = 'Creative & Media'

        elif any(x in text for x in [
            'security officer', 'security guard', 'loss prevention',
            'surveillance', 'law enforcement', 'investigator'
        ]):
            sector = 'Security & Protective'

        elif any(x in text for x in [
            'assembler', 'assembly line', 'machine operator', 'quality control',
            'lean manufacturing', 'six sigma', 'plant manager', 'production line'
        ]):
            sector = 'Manufacturing'

    return pd.Series({
        'sector_clean': sector,
        'seniority': seniority
    })


# Apply — axis=1 passes the full row
df[['sector_clean', 'seniority']] = df.apply(classify_job, axis=1)

print(df['sector_clean'].value_counts())
print(f"\nSector nulls remaining: {df['sector_clean'].isna().sum()}")
print(f"Seniority nulls:        {df['seniority'].isna().sum()}")
```

lets parse the location column

```python
def parse_location(val):
    if pd.isna(val): 
        return np.nan, np.nan
    # Remove ZIP codes
    val = re.sub(r'\s+\d{5}(-\d{4})?$', '', val.strip())
    parts = val.split(',')
    if len(parts) == 2:
        return parts[0].strip(), parts[1].strip()
    return val.strip(), np.nan

df[['city', 'state']] = pd.DataFrame(
    df['location'].apply(parse_location).tolist(),
    index=df.index
)

# Verify
print(df[['location', 'city', 'state']].head(10))
print(f"State nulls: {df['state'].isna().sum()}")
```
